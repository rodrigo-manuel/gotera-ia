```python

import os
import glob
import numpy as np
from PIL import Image
import tensorflow as tf

# Rutas de las imágenes extraídas
ruta_test = "/content/dataset_prueba/test"
carpetas = {
    "ojo_de_gallo": os.path.join(ruta_test, "ojo_de_gallo"),
    "sana": os.path.join(ruta_test, "sana")
}

# Comprobar que existen las imágenes
for clase, carpeta in carpetas.items():
    imagenes = glob.glob(os.path.join(carpeta, "*"))
    imagenes = [
        p for p in imagenes
        if p.lower().endswith((".jpg", ".jpeg", ".png", ".bmp"))
    ]
    print(f"{clase}: {len(imagenes)} imágenes")

# Cargar el intérprete TFLite
ruta_tflite = "/content/modelo_ojo_de_gallo.tflite"
interpreter = tf.lite.Interpreter(model_path=ruta_tflite)
interpreter.allocate_tensors()

entrada_info = interpreter.get_input_details()[0]
salida_info = interpreter.get_output_details()[0]

print("\nEntrada TFLite:", entrada_info["shape"], entrada_info["dtype"])
print("Salida TFLite:", salida_info["shape"], salida_info["dtype"])

# Comparar las predicciones con las mismas imágenes
resultados = []

for clase, carpeta in carpetas.items():
    imagenes = sorted([
        p for p in glob.glob(os.path.join(carpeta, "*"))
        if p.lower().endswith((".jpg", ".jpeg", ".png", ".bmp"))
    ])[:10]

    for ruta in imagenes:
        imagen = Image.open(ruta).convert("RGB").resize((224, 224))
        x = np.asarray(imagen, dtype=np.float32)[None, ...]

        # Modelo original Keras
        pred_keras = float(
            np.asarray(modelo(x, training=False)).reshape(-1)[0]
        )

        # Modelo convertido TFLite
        interpreter.set_tensor(entrada_info["index"], x)
        interpreter.invoke()
        pred_tflite = float(
            interpreter.get_tensor(salida_info["index"]).reshape(-1)[0]
        )

        resultados.append({
            "clase_real": clase,
            "archivo": os.path.basename(ruta),
            "keras": pred_keras,
            "tflite": pred_tflite,
            "diferencia": abs(pred_keras - pred_tflite)
        })

if resultados:
    diferencias = [r["diferencia"] for r in resultados]

    print(f"\nImágenes comparadas: {len(resultados)}")
    print(f"Diferencia máxima: {max(diferencias):.10f}")
    print(f"Diferencia media: {np.mean(diferencias):.10f}")

    print("\nPrimeros resultados:")
    for r in resultados[:10]:
        print(
            f'{r["clase_real"]:14s} | {r["archivo"][:30]:30s} | '
            f'Keras={r["keras"]:.6f} | '
            f'TFLite={r["tflite"]:.6f} | '
            f'Dif.={r["diferencia"]:.8f}'
        )
else:
    print("No se encontraron imágenes para comparar.")
```

```python

import os
import glob
import numpy as np
from PIL import Image
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score
)

ruta_test = "/content/dataset_prueba/test"

etiquetas_reales = []
etiquetas_predichas = []
probabilidades_sana = []

# Convención provisional:
# salida >= 0.5 -> sana
# salida < 0.5  -> ojo_de_gallo
# Debemos confirmar esta convención con el código de entrenamiento.

for clase in ["ojo_de_gallo", "sana"]:
    carpeta = os.path.join(ruta_test, clase)

    imagenes = [
        p for p in glob.glob(os.path.join(carpeta, "*"))
        if p.lower().endswith((".jpg", ".jpeg", ".png", ".bmp"))
    ]

    for ruta in imagenes:
        imagen = Image.open(ruta).convert("RGB").resize((224, 224))
        x = np.asarray(imagen, dtype=np.float32)[None, ...]

        salida = float(
            np.asarray(modelo(x, training=False)).reshape(-1)[0]
        )

        predicha = "sana" if salida >= 0.5 else "ojo_de_gallo"

        etiquetas_reales.append(clase)
        etiquetas_predichas.append(predicha)
        probabilidades_sana.append(salida)

orden = ["ojo_de_gallo", "sana"]

print("Total de imágenes:", len(etiquetas_reales))
print("Exactitud:", round(
    accuracy_score(etiquetas_reales, etiquetas_predichas), 4
))
print("\nMatriz de confusión:")
print("Orden de filas y columnas:", orden)
print(confusion_matrix(
    etiquetas_reales, etiquetas_predichas, labels=orden
))

print("\nInforme por clase:")
print(classification_report(
    etiquetas_reales,
    etiquetas_predichas,
    labels=orden,
    digits=4,
    zero_division=0
))
```

```python

import os
import numpy as np
from PIL import Image
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score
)

ruta_test = "/content/dataset_prueba/test"
clases = ["ojo_de_gallo", "sana"]

y_real = []
y_pred = []
errores = []

entrada_info = interpreter.get_input_details()[0]
salida_info = interpreter.get_output_details()[0]

for clase in clases:
    carpeta = os.path.join(ruta_test, clase)

    for nombre in sorted(os.listdir(carpeta)):
        if not nombre.lower().endswith((".jpg", ".jpeg", ".png", ".bmp")):
            continue

        ruta = os.path.join(carpeta, nombre)

        try:
            imagen = Image.open(ruta).convert("RGB").resize((224, 224))
            x = np.asarray(imagen, dtype=np.float32)[None, ...]

            interpreter.set_tensor(entrada_info["index"], x)
            interpreter.invoke()
            salida = float(
                interpreter.get_tensor(salida_info["index"]).reshape(-1)[0]
            )

            predicha = "sana" if salida >= 0.5 else "ojo_de_gallo"

            y_real.append(clase)
            y_pred.append(predicha)

            if predicha != clase:
                errores.append((nombre, clase, predicha, salida))

        except Exception as e:
            print("Error procesando", nombre, ":", str(e))

print("\nTotal evaluado:", len(y_real))
print("Exactitud:", round(accuracy_score(y_real, y_pred), 4))
print("\nMatriz de confusión (filas=real, columnas=predicción):")
print("Orden:", clases)
print(confusion_matrix(y_real, y_pred, labels=clases))

print("\nInforme por clase:")
print(classification_report(
    y_real, y_pred,
    labels=clases,
    digits=4,
    zero_division=0
))

print("Cantidad de imágenes clasificadas incorrectamente:", len(errores))
print("\nPrimeros errores (archivo, clase real, predicción, salida):")
for error in errores[:15]:
    print(error)
```

```python
import zipfile, json

with zipfile.ZipFile("/content/mejor_modelo.keras") as z:
    print("Metadatos:", z.read("metadata.json").decode())
    cfg = json.loads(z.read("config.json"))

def recorrer(n):
    if isinstance(n, dict):
        c = n.get("config") if isinstance(n.get("config"), dict) else {}
        clase = n.get("class_name")
        if clase in ("Rescaling", "Normalization"):
            print("CAPA", clase, "| nombre:", c.get("name"), "| scale:", c.get("scale"), "| offset:", c.get("offset"))
        if clase == "Dense":
            print("CAPA Dense | nombre:", c.get("name"), "| unidades:", c.get("units"), "| activación:", c.get("activation"))
        if c.get("name") == "data_augmentation":
            print("CAPA data_augmentation contiene:", [l.get("class_name") for l in c.get("layers", [])])
        for v in n.values():
            recorrer(v)
    elif isinstance(n, list):
        for v in n:
            recorrer(v)

recorrer(cfg)
```
